## Chain rule and shared paths
For a=x² and y=a²+a, dy/dx=(2a+1)2x. Predict the derivative at x=2 before running.

```python
x = Scalar(2)
a = x*x
y = a*a + a
y.backward()
print("Shared-path gradient:", x.grad)
assert x.grad == 36
```

## Independent numerical check
Central differences estimate the slope using two forward evaluations. Change h to 1e-2 and 1e-10; explain approximation and roundoff errors.

```python
import math
def forward(v):
    return math.tanh(v*v + 0.3*v)
x = Scalar(0.4)
y = (x*x + 0.3*x).tanh()
y.backward()
h = 1e-6
numeric = (forward(x.value+h)-forward(x.value-h))/(2*h)
print("autodiff", x.grad, "finite difference", numeric)
assert abs(x.grad-numeric) < 1e-6
```
